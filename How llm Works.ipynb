```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END

class State(TypedDict):
  question: str
  topic: str
  answer: str

def understand_question(state):
    question = state["question"]

    return {
      "topic": "Data Engineering"
    }

def create_answer(state):
  return {
    "answer": f"This is a {state['topic']} question."
  }


# Initialize the StateGraph using our State schema
graph = StateGraph(State)

graph.add_node(
    "understand_question",
    understand_question
)

graph.add_node(
    "create_answer",
    create_answer
)

graph.add_edge(
    START,
    "understand_question"
)

graph.add_edge(
    "understand_question",
    "create_answer"
)

graph.add_edge(
    "create_answer",
    END
)

app = graph.compile()

result = app.invoke({
    "question": "what is Spark partitioning?",
    "answer": ""
})

print(result)

try:
  png_data = app.get_graph().draw_mermaid_png()
  with open("langgraph.png", "wb") as f:
    f.write(png_data)
except Exception as e:
  print(f"Could not render graph image: {e}")
```
